```python
import numpy as np
```

## Vector to vector multiplication

```python
arr=np.array([2,3,4,6])
arr
```

```python
arr2=np.array([1,0,1,2])
arr2
```

```python
def vector_to_vector_multiplication(arr1,arr2):
    assert arr1.shape[0]==arr2.shape[0], "number of columns are not matching(rows in math term)"

    n=arr.shape[0]
    result= 0
    for i in range(n):
        result += arr1[i]*arr2[i]
    return result

res = vector_to_vector_multiplication(arr,arr2)
res
```

```python
arr.dot(arr2)
```

```python
arr3=np.array([2,4,6,8,10])
arr3
```

```python
vector_to_vector_multiplication(arr,arr3)
```

## Matrix to vector multiplication

```python
u=np.array([[10,20,30,40],
           [5,7,9,1],
           [3,4,3,2]])
u
```

```python
u.dot(arr)
```

## Matrix to Matrix multiplication

```python
v=np.array([[1,2,3],
           [5,6,7],
           [9,10,11],
           [12,13,14]])
v
```

```python
def matrix_matrix_multi(mat1,mat2):
    assert mat1.shape[1]==mat2.shape[0], "number of columns are not matching(rows in math term)"

    n=mat1.shape[0]
    m=mat1.shape[1]
    result=np.zeros((n,m))
    for ind in range(n):
        result[:,ind]=matrix_vector_multi(mat1,mat2[:,ind])
    return result

matrix_matrix_multi(u,v)
```

```python
u.dot(v)
```

```python
v.dot(u)
```
